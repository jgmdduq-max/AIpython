### 문제 12: 종류별 평균 비교하기

### 문제 15: 새로운 값 계산하기
- 어떤 종류가 꽃받침 면적이 가장 큰가요?
- 어떤 종류가 꽃잎 면적이 가장 큰가요?

```python
# 면적 컬럼 생성 (길이 * 너비)
df['SepalArea'] = df['SepalLengthCm'] * df['SepalWidthCm']
df['PetalArea'] = df['PetalLengthCm'] * df['PetalWidthCm']

# 종류별 평균 면적 계산
area_means = df.groupby('Species')[['SepalArea', 'PetalArea']].mean()
display(area_means)

# 결과 확인
max_sepal_area_species = area_means['SepalArea'].idxmax()
max_petal_area_species = area_means['PetalArea'].idxmax()

print(f"\n- 꽃받침 면적이 가장 큰 종류: {max_sepal_area_species} ({area_means.loc[max_sepal_area_species, 'SepalArea']:.2f} cm²)")
print(f"- 꽃잎 면적이 가장 큰 종류: {max_petal_area_species} ({area_means.loc[max_petal_area_species, 'PetalArea']:.2f} cm²)")
```

### 문제 16: 순위 매기기
- **Top 10의 기준**: 꽃잎 너비 + 꽃잎 길이 + 꽃받침 너비 + 꽃받침 길이의 합
- top 10에 가장 많이 포함된 종류는?
- 가장 적게 포함된 종류는?

```python
# 4가지 수치의 합계 컬럼 생성
df['TotalSize'] = df['SepalLengthCm'] + df['SepalWidthCm'] + df['PetalLengthCm'] + df['PetalWidthCm']

# TotalSize 기준 내림차순 정렬 후 상위 10개 추출
top_10_sum_df = df.sort_values(by='TotalSize', ascending=False).head(10)

display(top_10_sum_df[['Id', 'Species', 'TotalSize']])

# 상위 10개 내 종별 빈도 확인
top_10_sum_counts = top_10_sum_df['Species'].value_counts()
print("\n[합계 기준 상위 10개 데이터의 종별 분포]")
print(top_10_sum_counts)
```

### 문제 16-1: 가중치 적용 순위 매기기
- 각 항목(SepalLength, SepalWidth, PetalLength, PetalWidth)의 평균값을 구합니다.
- 각 수치를 해당 평균값으로 나누어 더한 `WeightedSum` 컬럼을 만듭니다.
- 이 값을 기준으로 Top 10을 뽑아 종별 분포를 확인합니다.

```python
# 1. 각 항목의 평균 계산
means = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']].mean()

# 2. 가중치 적용 (수치 / 평균) 하여 합산
# 각 항목이 1.0을 기준으로 기여하게 되어 수치 크기에 상관없이 고르게 반영됩니다.
df['WeightedSum'] = (
    (df['SepalLengthCm'] / means['SepalLengthCm']) +
    (df['SepalWidthCm'] / means['SepalWidthCm']) +
    (df['PetalLengthCm'] / means['PetalLengthCm']) +
    (df['PetalWidthCm'] / means['PetalWidthCm'])
)

# 3. WeightedSum 기준 상위 10개 추출
top_10_weighted_df = df.sort_values(by='WeightedSum', ascending=False).head(10)

display(top_10_weighted_df[['Id', 'Species', 'WeightedSum', 'SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']])

# 4. 종별 빈도 확인
weighted_counts = top_10_weighted_df['Species'].value_counts()
print("\n[가중치 적용 상위 10개 데이터의 종별 분포]")
print(weighted_counts)
```

### 문제 20: 품종별 데이터 분포 확인 (Box Plot)
- 각 수치형 열에 대해 품종별 분포와 이상치를 시각화합니다.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 2행 2열 구조로 박스플롯 그리기
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
titles = ['꽃받침 길이 분포', '꽃받침 너비 분포', '꽃잎 길이 분포', '꽃잎 너비 분포']

for i, col in enumerate(columns):
    row = i // 2
    col_idx = i % 2

    # hue를 지정하고 legend=False를 설정하여 경고 해결
    sns.boxplot(x='Species', y=col, hue='Species', data=df, ax=axes[row, col_idx],
                palette=['skyblue', 'orange', 'green'], legend=False)
    axes[row, col_idx].set_title(titles[i], fontsize=13)
    axes[row, col_idx].set_xlabel('품종')
    axes[row, col_idx].set_ylabel('길이/너비 (cm)')

plt.tight_layout()
plt.show()
```
