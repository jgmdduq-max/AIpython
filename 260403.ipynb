```python
import random
answer=random.randint(1,100)
```

```python
answer
```

```python
user=int(input("추측값을 입력하시오."))
```

```python
repeat=0
while user != answer:
  if user > answer:
    print("작은값을 입력하시오")
    repeat+=1
  else:
    print("큰값을 입력하시오.")
    repeat+=1
  user=int(input("추측값을입력하시오."))
print("총 시도 회수:", repeat)
```

```python
num=int(input("약수를 입력하시오."))
print(f"{num}의 약수는 다음과 같습니다:")
for i in range(1,num+1):
  if num%i==0:
    print(i)
```
