```python
cafe_menu_prices_en = {
    "Americano (Hot)": 4500,
    "Americano (Iced)": 5000,
    "Cafe Latte (Hot)": 5200,
    "Cafe Latte (Iced)": 5700,
    "Vanilla Latte": 5900,
    "Caramel Macchiato": 6200,
    "Cappuccino": 5500,
    "Cold Brew": 6000,
    "Hot Chocolate": 5800,
    "Grapefruit Ade": 6500
}

def show_menu():
    print("\n=== Cafe Menu ===")
    for name, price in cafe_menu_prices_en.items():
        print(f"- {name}: {price}원")
    print("\n※ 종료하려면 메뉴 입력에 '종료' 를 입력하세요.")

# 대소문자 무시 매칭용 (키: 소문자, 값: 원래 메뉴명)
menu_lookup = {name.lower(): name for name in cafe_menu_prices_en.keys()}

total_sum = 0

# 최종 결과에서 "어떤 메뉴를 몇 잔" 시켰는지 보여주기 위한 누적
order_summary = {}  # {메뉴명: 총 잔수}

while True:
    show_menu()
    user_input_raw = input("\n음료 종류를 입력하세요: ").strip()

    # '종료' 입력 시 프로그램 종료
    if user_input_raw.lower() == "종료":
        break

    user_input = user_input_raw.lower()

    # 메뉴 검증
    if user_input not in menu_lookup:
        print("\n메뉴에 없는 음료입니다. 다시 선택해주세요.")
        continue

    drink = menu_lookup[user_input]

    # 잔 수 입력
    while True:
        cups_str = input("잔 수(개수)를 입력하세요: ").strip()
        if cups_str.isdigit() and int(cups_str) >= 1:
            cups = int(cups_str)
            break
        print("잔 수는 1 이상의 정수로 입력해야 합니다. 다시 입력해주세요.")

    unit_price = cafe_menu_prices_en[drink]
    subtotal = unit_price * cups
    total_sum += subtotal

    # 메뉴별 잔 수 누적
    order_summary[drink] = order_summary.get(drink, 0) + cups

    print("\n=== 주문 추가 완료 ===")
    print(f"추가된 메뉴: {drink} / {cups}잔 (소계 {subtotal}원)")
    print(f"현재 누적 금액: {total_sum}원")

# 최종 결과 출력(메뉴별 잔 수가 한눈에 보이게)
print("\n====================")
print("   최종 주문 결과")
print("====================")

if not order_summary:
    print("주문한 내역이 없습니다.")
else:
    print("[메뉴별 주문 잔수]")
    # 보기 좋게 이름 순으로 정렬(원하면 삭제해도 됨)
    for drink in sorted(order_summary.keys()):
        cups = order_summary[drink]
        unit_price = cafe_menu_prices_en[drink]
        print(f"- {drink}: {cups}잔  (가격 {unit_price}원)")

print(f"\n[총 결제 금액] {total_sum}원")
```
